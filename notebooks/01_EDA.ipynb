# EstateAI - Exploratory Data Analysis (EDA)

## Objective

This notebook performs exploratory data analysis on the House Prices dataset.

Goals:

- Understand the dataset
- Identify missing values
- Analyze target variable distribution
- Explore feature relationships
- Prepare data for preprocessing and model training

```python
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 150)

print("Libraries loaded successfully!")
```

```python
DATA_PATH = Path("../data/raw/train.csv")

df = pd.read_csv(DATA_PATH)

print(df.shape)
```

```python
missing = (
    df.isnull()
      .sum()
      .sort_values(ascending=False)
)

missing = missing[missing > 0]

missing.to_frame("Missing Values")
```

```python
plt.figure(figsize=(10,5))

plt.hist(df["SalePrice"], bins=50)

plt.title("Sale Price Distribution")
plt.xlabel("Sale Price")
plt.ylabel("Count")

plt.show()
```

```python
corr = (
    df.corr(numeric_only=True)["SalePrice"]
      .sort_values(ascending=False)
)

corr.head(15)
```
